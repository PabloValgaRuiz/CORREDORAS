```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
charcoal = pd.read_csv('charcoal_age_v2.csv')[['ka.BP', 'CHAR']]
charcoal = charcoal[charcoal['ka.BP'] <= 13.7]

plt.figure(figsize=(16, 12))
charcoal.plot(x='ka.BP', y='CHAR', color='red')
# plot vertical lines on 11.1, 6.8, 2.2
plt.plot([11.1, 11.1], [0, charcoal['CHAR'].max()], color='black', linestyle='--', alpha=0.8)
plt.plot([6.8, 6.8], [0, charcoal['CHAR'].max()], color='black', linestyle='--', alpha=0.8)
plt.plot([2.2, 2.2], [0, charcoal['CHAR'].max()], color='black', linestyle='--', alpha=0.8)
plt.gca().invert_xaxis()
plt.xlabel('')
plt.legend([])
```
