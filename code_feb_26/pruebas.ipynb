```python
def coef_sp1_a_sp2_condicional(df, sp1, sp2):

    try:

        x = df[sp1].values
        y = df[sp2].values

        temp = df["Temperatura"]
        mult = df["Multiplier"]

        data = pd.DataFrame({
            sp1: x,
            sp2: y,
            "Temp": temp,
            "Mult": mult
        })

        model = VAR(data)
        results = model.fit(maxlags=1)

        A = results.coefs[0]

        # fila 1 = destino sp2
        # columna 0 = origen sp1
        return A[1][0]
    
    except:
        return 0
```
