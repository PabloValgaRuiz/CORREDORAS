```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return 0.7*x + np.random.normal(0, 0.1, size=x.shape)*100
```

```python
x = np.linspace(0, 10, 4000)
y = f(x)

plt.scatter(x, y, s=10, alpha=0.5)

# spearman
from scipy.stats import spearmanr
rho, pval = spearmanr(x, y)
plt.title(f"Spearman correlation: {rho:.2f}, p-value: {pval:.2e}")
plt.ylim(-50,50)
plt.grid()
plt.plot(x, 0.7*x, color='red', linewidth=2)
plt.show()
```
