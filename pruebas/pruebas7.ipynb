```python
def size_my_pov(obj_size, scale):
    return obj_size * scale
def scale_her_pov(obj_size, scale):
    return obj_size / scale
def set_scale_size(size = None, scale = None):
    og_size = 1.80
    if size is not None:
        scale = og_size / size
    elif scale is not None:
        size = og_size * scale
    else:
        raise ValueError("Either size or scale must be provided")
    return og_size, size, scale
```
